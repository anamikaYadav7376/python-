```python
i = np.random.randint(2,15,9)

j = np.random.randint(2,15,9)


k = np.concatenate((i,j))
print (k)

result = np.array([i,j])
print(result)

# Shapes: a is (2, 2), b is (1, 2)
a = np.array([[1, 2],
              [3, 4]])
b = np.array([[5, 6]])

# Goal: Append 'b' as a new column.
# Problem: We cannot concatenate (2, 2) and (1, 2) along axis=1.
# Solution: Transpose 'b' to shape (2, 1) first.
result = np.concatenate((a, b.T), axis=1)

print(result)
```

```python
import numpy as np
prices = []
items = ["Rice", "Sugar", "Milk", "Oil",
         "Tea", "Soap", "Biscuits"]
for item in items:
    price = float(input(f"Enter price of {item}: "))
    prices.append(price)
prices = np.array(prices)
print("Prices:", prices)
print("Max Price:", np.max(prices))
print("Min  Price:", np.min(prices))
print("Avg Price:", np.mean(prices))
print("Std dev:", np.std(prices))
```

```python
import numpy as np

matrix = np.random.randint(1, 91, (5,5))

print(matrix)

# Find mean of values row and column wise

print(np.mean(matrix, axis=1))  # row - wise
print(np.mean(matrix, axis=0) ) # column-wise

# Replace all even numbers with -1

# matrix[matrix % 2 == 0] = -1
# print(matrix)

# OR

modified_Mat = np.where(matrix % 2 == 0, -1, matrix)
print(modified_Mat)

# To Sort the array row and column wise

print(np.sort(matrix, axis=1))
print(np.sort(matrix, axis=0))

# Find row and column wise sum saparately

Row_sum = np.sum(matrix, axis=1)
print("Row sum : ", Row_sum)

Col_sum = np.sum(matrix, axis=0)
print("Col sum : ", Col_sum)

# count the "-ve" and where

# print(np.sum(matrix < 0, axis=0))

cnt = np.sum(np.where(modified_Mat < 0, 1 , 0))
print("Number of negative : ",cnt)
```

```python
stock = np.random.randint(30,100,(5,5))
print(stock)

demand = np.random.randint(1,100,(5,5))
print(demand)

remaining_stock = stock - demand
print(remaining_stock)

# Replace
replace = np.where(remaining_stock < 0, 0, remaining_stock)
print(replace)

# Find total shortage

total_shortage = np.sum(np.where(remaining_stock < 0, remaining_stock*-1, 0))
print(total_shortage)

# Product with zero stock

zero_stock_products = np.sum(np.where(replace==0,1,0), axis=0)
print(zero_stock_products)

# Sort each row based on remaining stocks

row_sort = np.sort(remaining_stock, axis=1)
print(row_sort)
```
