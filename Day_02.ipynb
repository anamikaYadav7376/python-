# LIST []

```python
lst = [1,2,"Hello",True,[2,3]]

print(lst)

# append()

lst.append(3)
print(lst)

#this will print a list if we give list

lst.append(["hi","bye"])
print(lst)
```

```python
#extend()

lst.extend(["Hi","Bye"]) # this with print separeted
print(lst)
```

```python
#insert()

lst.insert(2,"value")
print(lst)
```

```python
# pop - remove last element

lst.pop()
print(lst)
```

```python
# Remove - remove 1st occurance of a value

lst.remove(4)
print(lst)
```

```python
# index()

lst.index(2)
```

```python
# count()

lst.count(2)
```

```python
lst = [1,9,3,6,2]

# Sort

lst.sort(reverse=True)

print(lst)
```

```python
# These are FUNCTIONS
# they all create a new list

lst = [1,9,3,6,2,7]

# sum

print(sum(lst))

# min

print(min(lst))

# max

print(max(lst))

#sorted

print(sorted(lst))
```

```python
lst = [1,2,3,4,5]
lst2 = [6,7,8,9,0]

i=0
j=0
add = []

while (i<len(lst) and j<len(lst2)):
  if(i == j ):
    add.append(lst[i]+lst2[j])
  i+=1
  j+=1

print(add)
```

# split() -- breaks a string into smaller pieces and returns them as a list.

```python
lst = input().split()
lst2 = input().split()

for i in range(len(lst)):
    lst[i] = int(lst[i])

for i in range(len(lst2)):
    lst2[i] = int(lst2[i])

add = []

for i in range(len(lst)):
    add.append(lst[i] + lst2[i])

print(add)
```
